```python
# Forward Propagation
import numpy as np
import pandas as pd

data = {'cgpa': [8.5, 9.2, 7.8], 'profile_score': [85, 92, 78], 'lpa': [10, 12, 8]}
df = pd.DataFrame(data)
X = df[['cgpa', 'profile_score']].values

def initialize_parameters():
    np.random.seed(1)
    W = np.random.randn(2, 1) * 0.01
    b = np.zeros((1, 1))
    return W, b
def forward_propagation(X, W, b):
    Z = np.dot(X, W) + b
    A = 1 / (1 + np.exp(-Z))
    return A
W, b = initialize_parameters()
A = forward_propagation(X, W, b)
print("Final Output:", A)
```

### Giải thích Code và Nhận xét Kết quả

#### 1. Giải thích Code
* **Khởi tạo dữ liệu**: Dữ liệu đầu vào `X` gồm hai thuộc tính `cgpa` và `profile_score` từ 3 mẫu dữ liệu.
* **Khởi tạo tham số (`initialize_parameters`)**:
  * Ma trận trọng số $W$ được khởi tạo ngẫu nhiên theo phân phối chuẩn với kích thước `(2, 1)` và được nhân với `0.01` để giữ giá trị nhỏ, giúp mô hình hội tụ tốt hơn.
  * Hệ số chệch $b$ được khởi tạo bằng `0` với kích thước `(1, 1)`.
* **Lan truyền xuôi (`forward_propagation`)**:
  * Tính tổng tuyến tính: $Z = X \cdot W + b$.
  * Áp dụng hàm kích hoạt Sigmoid: $A = \sigma(Z) = \frac{1}{1 + e^{-Z}}$ để chuyển đổi kết quả thành giá trị trong khoảng $(0, 1)$.
