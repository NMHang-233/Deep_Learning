### 1. Giới thiệu bài toán và Chuẩn bị dữ liệu

Trong phần này, chúng ta tiến hành tải tập dữ liệu phân loại Chó và Mèo (`cat-vs-dog-dataset`) từ Kaggle thông qua thư viện `kagglehub`.

* **Tự động tìm đường dẫn:** Sử dụng hàm `find_image_dirs` để quét cấu trúc thư mục được tải xuống nhằm tìm chính xác vị trí chứa ảnh chó và mèo.
* **Trực quan hóa:** Hiển thị một số ảnh mẫu ngẫu nhiên của chó và mèo để kiểm tra dữ liệu đầu vào.

```python
#Cat and Dog
#Cat and Dog
import os
import warnings
from zipfile import ZipFile
import kagglehub  # Thêm kagglehub để tải dữ liệu

import matplotlib.image as mpimg
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import tensorflow as tf
from keras import layers
from tensorflow import keras
from tensorflow.keras.layers import (
    Activation,
    BatchNormalization,
    Conv2D,
    Dense,
    Dropout,
    Flatten,
    MaxPooling2D,
)
from tensorflow.keras.models import Sequential
from tensorflow.keras.preprocessing import image, image_dataset_from_directory
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img

warnings.filterwarnings('ignore')

# --- 1. Tải dữ liệu bằng Kagglehub ---
# Download latest version
path = kagglehub.dataset_download("anubhab098/cat-vs-dog-dataset")
print("Path to dataset files:", path)

# Tự động tìm đường dẫn chứa các thư mục con cat/dog thực tế (đề phòng thư mục lồng nhau)
def find_image_dirs(base_path):
    for root, dirs, files in os.walk(base_path):
        lowercased_dirs = [d.lower() for d in dirs]
        has_cat = any('cat' in d for d in lowercased_dirs)
        has_dog = any('dog' in d for d in lowercased_dirs)
        if has_cat and has_dog:
            cat_folder = [d for d in dirs if 'cat' in d.lower()][0]
            dog_folder = [d for d in dirs if 'dog' in d.lower()][0]
            return os.path.join(root, cat_folder), os.path.join(root, dog_folder), root
    return None, None, base_path

cat_dir, dog_dir, base_dir = find_image_dirs(path)

if not cat_dir or not dog_dir:
    raise FileNotFoundError(f"Không tìm thấy thư mục cat và dog trong {path}. Hãy kiểm tra cấu trúc: {os.listdir(path)}")

print("Tìm thấy thư mục mèo tại:", cat_dir)
print("Tìm thấy thư mục chó tại:", dog_dir)
print("Thư mục gốc dataset thực tế (base_dir):", base_dir)

# --- 2. Trực quan hóa dữ liệu (Hiển thị một số ảnh mẫu) ---
fig = plt.gcf()
fig.set_size_inches(16, 16)

cat_names = os.listdir(cat_dir)
dog_names = os.listdir(dog_dir)

pic_index = min(210, len(cat_names), len(dog_names))
cat_images = [
    os.path.join(cat_dir, fname)
    for fname in cat_names[max(0, pic_index - 8) : pic_index]
]
dog_images = [
    os.path.join(dog_dir, fname)
    for fname in dog_names[max(0, pic_index - 8) : pic_index]
]

for i, img_path in enumerate(cat_images + dog_images):
  sp = plt.subplot(4, 4, i + 1)
  sp.axis('Off')
  img = mpimg.imread(img_path)
  plt.imshow(img)
plt.show()

# --- 3. Chia tập dữ liệu Train / Validation ---
train_datagen = image_dataset_from_directory(
    base_dir,
    image_size=(200, 200),
    subset='training',
    seed=1,
    validation_split=0.1,
    batch_size=32,
)

test_datagen = image_dataset_from_directory(
    base_dir,
    image_size=(200, 200),
    subset='validation',
    seed=1,
    validation_split=0.1,
    batch_size=32,
)

# --- 4. Xây dựng mô hình CNN ---
model = tf.keras.models.Sequential([
    layers.Conv2D(
        32, (3, 3), activation='relu', input_shape=(200, 200, 3)
    ),
    layers.MaxPooling2D(2, 2),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D(2, 2),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D(2, 2),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D(2, 2),
    layers.Flatten(),
    layers.Dense(512, activation='relu'),
    layers.BatchNormalization(),
    layers.Dense(512, activation='relu'),
    layers.Dropout(0.1),
    layers.BatchNormalization(),
    layers.Dense(512, activation='relu'),
    layers.Dropout(0.2),
    layers.BatchNormalization(),
    layers.Dense(1, activation='sigmoid'),
])

model.summary()

# --- 5. Biên dịch và Huấn luyện mô hình ---
model.compile(
    loss='binary_crossentropy', optimizer='adam', metrics=['accuracy']
)

history = model.fit(train_datagen, epochs=10, validation_data=test_datagen)

# --- 6. Đánh giá mô hình (Vẽ biểu đồ Loss & Accuracy) ---
history_df = pd.DataFrame(history.history)
history_df.loc[:, ['loss', 'val_loss']].plot()
history_df.loc[:, ['accuracy', 'val_accuracy']].plot()
plt.show()


# --- 7. Hàm dự đoán ảnh mới ---
def predict_image(image_path):
  img = image.load_img(image_path, target_size=(200, 200))
  plt.imshow(img)
  plt.show()

  img_array = image.img_to_array(img)
  img_array = np.expand_dims(img_array, axis=0)

  result = model.predict(img_array)
  print('Dog' if result >= 0.5 else 'Cat')


# Chạy thử nghiệm dự đoán với ảnh trong dataset vừa tải
predict_image(os.path.join(cat_dir, cat_names[0]))
predict_image(os.path.join(dog_dir, dog_names[0]))
```

### 2. Thiết lập tập dữ liệu Train / Validation và Xây dựng cấu trúc mô hình CNN

* **Chia tập dữ liệu:** Sử dụng `image_dataset_from_directory` để chia dữ liệu thành tập huấn luyện (Train - 90%) và tập kiểm định (Validation - 10%) với kích thước ảnh thống nhất là `200x200` pixel.
* **Kiến trúc mạng CNN:**
    * **Khối Convolutional (Conv2D + MaxPooling2D):** 4 lớp trích xuất đặc trưng với số bộ lọc tăng dần từ 32 đến 64 giúp nắm bắt các họa tiết, góc cạnh và hình dáng của vật thể.
    * **Lớp phẳng hóa (Flatten) & Khối kết nối đầy đủ (Dense):** Chuyển đổi ma trận đặc trưng thành vector 1 chiều và đưa qua các lớp Dense có kích thước 512.
    * **Batch Normalization & Dropout:** Giúp ổn định quá trình tối ưu, tăng tốc độ hội tụ và giảm thiểu hiện tượng quá khớp (Overfitting).
    * **Hàm kích hoạt đầu ra:** Sử dụng `sigmoid` ở lớp cuối cùng phù hợp cho bài toán phân loại nhị phân (Cat: 0, Dog: 1).

### 3. Đánh giá và Nhận xét Kết quả Huấn luyện

Dựa trên biểu đồ biểu diễn giá trị Loss (Độ lỗi) và Accuracy (Độ chính xác) qua các epoch, ta có các nhận xét sau:

1. **Hiệu năng của mô hình:**
   * Kiểm tra mức độ hội tụ của đường `loss` và `val_loss`. Nếu `val_loss` giảm và đi ngang cùng `loss`, mô hình đang học tốt.
   * Kiểm tra khoảng cách giữa `accuracy` (trên tập huấn luyện) và `val_accuracy` (trên tập kiểm định). Khoảng cách này nhỏ chứng tỏ mô hình có khả năng tổng quát hóa tốt.
2. **Hiện tượng Overfitting:**
   * Nếu đường `loss` tiếp tục giảm mạnh nhưng `val_loss` lại có xu hướng đi lên sau một số epoch, mô hình đang gặp hiện tượng quá khớp (overfitting). Việc thêm `Dropout` và `BatchNormalization` trong kiến trúc trên đã phần nào hạn chế vấn đề này.
3. **Thử nghiệm dự đoán:**
   * Hàm `predict_image` chuyển đổi ảnh mới về kích thước `200x200`, đưa qua mô hình dự đoán và trả về nhãn cụ thể dựa trên ngưỡng xác suất `0.5`.
